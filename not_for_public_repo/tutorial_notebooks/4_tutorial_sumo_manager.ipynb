# Tutorial: SUMO Manager
## 1. Introduction to SUMO manager.
The [SUMO-manager](https://gitlab.lrz.de/cps/commonroad-sumo-manager/-/tree/development) is a module contains a complete CommonRoad-SUMO interface. The SUMO simulator is executed in a Docker container, which is started automatically by the interface. The aim of the SUMO-manager is to help the user convert, generate or simulate a scenario without installing so many external repositories. 

The module is still under development, if you encounter any issue or you have some feature requests, feel free to contact us! :)

## 2. Utilize SUMO manager in scenario conversion.

The functionality of converting scenario using SUMO-manager is not yet implemented in this repository. So we need to import the function **convert_scenario** from SUMO-manager.

```python
from configuration import CONFIG_TYPE
from commonroad-sumo-manager.example.convert_scenario import convert_scenario

scenario_name = "DEU_A9-2_1_T-1"
input_scenario_folder = "./example_scenarios/cr_scenario"
output_scenario_folder = "./example_scenarios/output/tutorial_4"

convert_scenario(scenario_name  = scenario_name,
                input_scenario_folder = input_scenario_folder,
                output_scenario_folder = output_scenario_folder)
```

## 3. Utilize SUMO manager in scenario simulation.

The utilization of SUMO-manager in simulating a scenario is easier as it is already integrated in the interactive benchmark. You can call that by switching the argument **use_sumo_manager** of the function **simulate_interactive_solution** to True.


```python
from evaluation.simulate_solution import simulate_interactive_solution

interactive_scenario_folder = "./example_scenarios/output/tutorial_4/DEU_A9-2_1_I-1-1"
solution_file = "./example_scenarios/solution/KS1:SA1:DEU_A9-2_1_T-1:2018b.xml"
output_folder_path = "./example_scenarios/gif"
creating_video = True 
use_sumo_manager = True

without_ego, with_ego = simulate_interactive_solution(interactive_scenario_folder = interactive_scenario_folder,
                              solution_file = solution_file,
                              output_folder_path = output_folder_path,
                              creating_video = creating_video,
                              use_sumo_manager = use_sumo_manager)
```
