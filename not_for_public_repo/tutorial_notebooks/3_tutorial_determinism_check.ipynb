# Tutorial: Determinism Check on Interactive Scenarios
To ensure that the simulations always produce the same results for a given trajectory of ego vehicle, it is important that we have a determinism check. Here we show how this is implemented. The check consist of two parts:
* **simulation_checker**: re-simulates the scenario for multiple times and compares the simulation results.
* **route_file_checker**: checks if any key parameters are set to **'random'** in the **.rou.xml** file.

The function **check_determinism()** calls this two checkers. If you wish to run these checks, the following arguments are required:
* **scenario_folder_path**: path to the folder containing required SUMO network files for the sceanrio. 
* **output_folder_path**: path to store the animation files of the simulation. 
* **creating_video**: a boolean value indicating whether a video file of the SUMO simulation should be created or not. **Default value**: False.
* **num_of_simulations**: The number of resimulation. **default**: 5
* **sumo_manager**: a boolean value indicating whether [SUMO Manager](https://gitlab.lrz.de/cps/commonroad-sumo-manager) should be used. **Default value**: False.

We first show how the two checkers work individually.

```python
from determinism_checker.route_file_checker import rou_file_determinism_check
from determinism_checker.simulation_checker import resimulate_scenario, plot_vehicle_trajectories, \
    obstacle_determinism_check
from determinism_checker.check_determinism import check_determinism

# prepare required arguments
scenario_file_path = "./example_scenarios/output/tutorial_2/DEU_A9-2_1_I-1-1"
num_of_simulations = 5
output_folder_path = "./determinism_check"
use_sumo_manager = False
creating_video = False

simulated_scenarios = resimulate_scenario(scenario_folder_path = scenario_file_path, 
                                          num_of_simulations = num_of_simulations,
                                          use_sumo_manager = use_sumo_manager,
                                          creating_video = creating_video,
                                          output_folder_path = output_folder_path)

is_vehicle_deterministic = obstacle_determinism_check(simulated_scenarios)
is_rou_files_deterministic = rou_file_determinism_check(scenario_file_path + 
                                                        "/DEU_A9-2_1_I-1-1.vehicles.rou.xml")
print(f"Determinism check result: {is_vehicle_deterministic and is_rou_files_deterministic}")
```

But in practice, just calling the **check_determinism** function can check the determinism of the scenario.

```python
result_check = check_determinism(scenario_file_path = scenario_file_path,
                                 num_of_simulations = num_of_simulations,
                                 output_folder_path = output_folder_path,
                                 use_sumo_manager = use_sumo_manager,
                                 creating_video = creating_video)
print(f"Determinism check result: {result_check}")
```
