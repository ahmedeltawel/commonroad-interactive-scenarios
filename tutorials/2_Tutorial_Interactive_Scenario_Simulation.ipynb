# Tutorial: Simulating an Interactive Scenario

This tutorial shows how to simulate a successfully converted interactive scenario with and without a given trajectory of the ego vehicle. You can see that the motion of obstacle vehicles varies when ego vehicle joins the scenario.

This block runs the same conversion as what you did in tutorial 1, you can skip this part if you have already gone through that tutorial and successfully ran all the codes with the expected outputs..

```python
%matplotlib inline

%load_ext autoreload
%autoreload 2

import sys
sys.path.append("..")
#Converting the same static CR scenario as done in tutorial 1 again.
from scenario_converter.convert_scenarios import convert_scenario
from configuration import CONFIG_TYPE

cr_scenario_path  = "../example_scenarios/cr_scenario/DEU_A9-2_1_T-1.xml"
output_folder_path = "../example_scenarios/output/tutorial_2"
config_type = CONFIG_TYPE.SUMO_CONFIG_1
creating_video = False

convert_scenario(cr_scenario_path  = cr_scenario_path,
                output_folder_path = output_folder_path,
                config_type = config_type,
                creating_video = creating_video)
```

The simulation is implemented by the function **simulate_solution** with the following arguments:
* **interactive_scenario_folder**: The path of input interactive scenario folder. 
* **solution_file**: The path of the CommonRoad solution file, where the trajectory of ego vehicle is extrated. 
* **output_folder_path**: The path where the generated gif is saved. 
* **creating_video**: Boolean value determining whether to generate a gif or not. **default**: False
* **use_sumo_manager**: [SUMO-manager](https://gitlab.lrz.de/cps/commonroad-sumo-manager) is a manager for dockerized SUMO simulator, with this tool we can implement many functions without installing SUMO and CommonRoad-SUMO-interface. A detailed introduction to the usage of SUMO-manager can be found in tutorial 5. **default**: False

```python
from evaluation.simulate_solution import simulate_interactive_solution

interactive_scenario_folder = "../example_scenarios/output/tutorial_2/DEU_A9-2_1_T-1-1"
solution_file = "../example_scenarios/solution/KS1:SA1:DEU_A9-2_1_T-1:2018b.xml"
output_folder_path = "../example_scenarios/gif"
creating_video = True 
use_sumo_manager = False

without_ego, with_ego = simulate_interactive_solution(interactive_scenario_folder = interactive_scenario_folder,
                              solution_file = solution_file,
                              output_folder_path = output_folder_path,
                              creating_video = creating_video,
                              use_sumo_manager = use_sumo_manager)
```

Now we can compare the results from the simulation. You can see the simulation with the trajectory of the ego vehicle on the left and without the trajectory on the right. It's obvious that during the simulation, the trajectory of obstacle vehicles are changed due to the addition of ego vehicle.
