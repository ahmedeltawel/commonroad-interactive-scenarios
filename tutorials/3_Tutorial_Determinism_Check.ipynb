# Tutorial: Check the Determinism of an Interactive Scenario
Checking if an interactive scenario is deterministic or not is implemented in two ways:
* **simulation_checker** re-simulates the scenario for multiple times and compare the results.
* **route_file_checker** checks if any key parameters are set to **'random'** in the **.rou.xml** file.

But you don't need to run them separately, we integrated the two checker into the **check_determinism**. Running that requires the following arguments:

* **scenario_folder_path**: The path of input interactive scenario folder.
* **output_folder_path**: The path where the generated gif is saved. 
* **creating_video**: Boolean value determining whether to generate a gif or not. **default**: False
* **num_of_simulations**: The number of resimulation. **default**: 5
* **sumo_manager**: Boolean value determining whether to use SUMO-manager or not **default**: False

First, we want to show that how the both functions work seperately.

```python
import sys
sys.path.append("..")
from determinism_checker.route_file_checker import rou_file_determinism_check
from determinism_checker.simulation_checker import resimulate_scenario, plot_vehicle_trajectories, \
    obstacle_determinism_check
from determinism_checker.check_determinism import check_determinism

scenario_file_path = "../example_scenarios/output/tutorial_2/DEU_A9-2_1_T-1-1"
num_of_simulations = 5
output_folder_path = "../determinism_check"
use_sumo_manager = False
creating_video = False

simulated_scenarios = resimulate_scenario(scenario_folder_path = scenario_file_path, 
                                          num_of_simulations = num_of_simulations,
                                          use_sumo_manager = use_sumo_manager,
                                          creating_video = creating_video,
                                          output_folder_path = output_folder_path)

is_vehicle_deterministic = obstacle_determinism_check(simulated_scenarios)

is_rou_files_deterministic = rou_file_determinism_check(scenario_file_path + 
                                                        "/DEU_A9-2_1_T-1-1.vehicles.rou.xml")
print(is_vehicle_deterministic and is_rou_files_deterministic)

```

But in practice, just calling the **check_determinism** function can check the determinism of the scenario.

```python
check_determinism(scenario_file_path = scenario_file_path,
                  num_of_simulations = num_of_simulations,
                  output_folder_path = output_folder_path,
                  use_sumo_manager = use_sumo_manager,
                  creating_video = creating_video)
```
